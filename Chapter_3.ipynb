## 3 The Four Fundamental Subspaces

### 3.1 Vector Spaces and Subspaces

$$\boxed{
\begin{aligned}
1.\ &\text{Requirement: All linear combinations } cv + dw \text{ must stay in the vector space.} \\
2.\ &\text{The row space of } A \text{ is “spanned” by the rows of } A. \text{The columns of } A \text{ span } C(A). \\
3.\ &\text{Matrices } M_1 \text{ to } M_N \text{ and functions } f_1 \text{ to } f_N \text{span matrix spaces and function spaces.}
\end{aligned}
}$$

#### Subspaces of Vector Spaces

- **DEFINITION**

    A subspace of a vector space is a set of vectors (including $\mathbf 0$) that satisfies two requirements: If $\mathbf v$ and $\mathbf w$ are vectors in the subspace and $c$ is any scalar, then

    1. $\mathbf v+\mathbf w$ is in the subspace

    2. $c\mathbf v$ is in the subspace

    3. Every subspace contains the zero vector.

    Another subspace is all of $\mathbb R^3$. The whole space is a subspace (of itself). Here is a list of all the possible subspaces of $\mathbb R^3$:

    **A subspace containing $\mathbf v$ and $\mathbf w$ must contain all linear combinations $c\mathbf v + d\mathbf w$.**

    $$\boxed{
    \begin{aligned}
    &(L)\text{ Any line through } (0,0,0) 
    \qquad
    &&(R^3)\text{ The whole space} \\
    &(P)\text{ Any plane through } (0,0,0)
    \qquad
    &&(Z)\text{ The single vector } (0,0,0)
    \end{aligned}
    }$$

    #### Example 1: First quadrant ($x \geq 0, y \geq 0$)

    * The set includes vectors with **nonnegative components**.
    * Example: $(2, 3)$ is included, but $(−2, −3)$ is not.
    * If we multiply $(2, 3)$ by −1, we get $(−2, −3)$, which is **not in the set**.

    **Conclusion:**  
    - Not closed under scalar multiplication $\to$ **Not a subspace**.

    #### Example 2: First and third quadrants  
    (vectors with both components $ \geq 0$ or both $\leq 0$)

    * Now scalar multiplication works (multiplying by any scalar stays inside).
    * But vector addition fails:

    * $(2, 3) + (−3, −2) = (−1, 1)$
    * This result is in **neither quadrant**.

    **Conclusion:**  
    - Not closed under addition $\to$ **Not a subspace**.

    #### Example 3: Subspaces of 2×2 matrices

    We work inside the vector space **$M$** of all 2×2 matrices.

    ##### Two important subspaces:

    1. Upper triangular matrices $U$

        $$
        U = \left\{
        \begin{bmatrix}
        a & b \\
        0 & d
        \end{bmatrix}
        \right\}
        $$

        * Closed under addition: sum of two upper triangular matrices is still upper triangular.
        * Closed under scalar multiplication.
        * Contains the zero matrix.

        So **$U$ is a subspace of $M$**.

    2. Diagonal matrices $D$

        $$
        D = \left\{
        \begin{bmatrix}
        a & 0 \\
        0 & d
        \end{bmatrix}
        \right\}
        $$

        * Closed under addition.
        * Closed under scalar multiplication.
        * Contains the zero matrix.

        So **$D$ is a subspace of $M$**.

    3. Relationship between them

        $$
        D \subset U \subset M
        $$

        Every diagonal matrix is also upper triangular, so **$D$ is a subspace of $U$**.

    4. Other subspaces mentioned

        * **Zero subspace $Z$**: only the zero matrix $\to$ always a subspace.
        * **Multiples of the identity**:
        $$
        cI = c
        \begin{bmatrix}
        1 & 0 \\
        0 & 1
        \end{bmatrix}
        $$
        These form a subspace of $M$ (also inside $U$ and $D$).

#### The Column Space of $A$

The **column space of a matrix $A$**, denoted $C(A)$, is the set of **all linear combinations of the columns of $A$**. In other words, it is the set of all vectors that can be written in the form

$$
Ax
$$

for some vector $ x $.

* Solving the system $ Ax = b $ means expressing $ b $ as a **combination of the columns of $A$**.

* Therefore:

  **$ Ax = b $ has a solution if and only if $ b \in C(A) $.**

* The column space is a **vector space**:
  * It contains the zero vector.
  * It is closed under addition and scalar multiplication.
  * It consists of all possible outputs of the linear transformation $ x \mapsto Ax $.

Important caution:

* The **columns themselves alone do NOT form a subspace**.
* Only **all linear combinations of them** form a subspace.
* We say the columns of $A$ **span** the column space.

Intuition:

- The column space represents **all vectors that matrix $A$ can reach**.  
- If $ b $ is outside this space, then no choice of $ x $ can solve $ Ax = b $.

### 3.2 Computing the Nullspace by Elimination: $A=CR$

$$\boxed{
\begin{aligned}
&\textbf{1. } \mathcal{N}(A) \subset \mathbb{R}^n \text{ contains all solutions } x \text{ to } Ax = 0, \text{ including } x = 0. \\

&\textbf{2. } \text{Elimination from } A \text{ to } R_0 \text{ to } R \text{ does not change the nullspace: }  \mathcal{N}(A) = \mathcal{N}(R_0) = \mathcal{N}(R). \\

&\textbf{3. } R_0 = \operatorname{rref}(A) \text{ has } I \text{ in } r \text{ columns and } F \text{ in } n-r \text{ columns.} \\

&\textbf{4. } \text{If column } j \text{ is dependent on previous columns, } Ax=0 \text{ has a special solution } \text{with } x_j = 1. \\

&\textbf{5. } \text{The } n-r \text{ special solutions to } Ax=0 \text{ contain } -F \text{ and } I. \\

&\textbf{6. } \text{Every short wide matrix } (m<n) \text{ has nonzero solutions to } Ax=0 \text{ in its nullspace.}
\end{aligned}
}$$

With square invertible matrices, Chapter 2 simplified $A$ to an upper tnangular $U$. For matrices of all shapes, in this chapter, elimination will simplify $Ax = 0$ to an “echelon form” $Rx = 0$.

For any matrix $A$ with rank $r$, the goal is to factor $A$ as

$$
A = CR
$$

where:
- the columns of **$C$** form a basis for the **column space** of $A$,
- the rows of **$R$** form a basis for the **row space** of $A$.

The matrix $C$: independent columns
- $C$ consists of the **first $r$ linearly independent columns of $A$** (pivot columns).
- These columns are taken **directly from $A$**.
- They span the column space of $A$.

The matrix $R$: column relationships
- Every remaining column of $A$ is a **linear combination of the columns of $C$**.
- These dependencies are encoded in a matrix $F$.
- The row factor has the form
$$
R = [\, I \;\; F \,] P
$$
where:
- $I$ corresponds to independent columns,
- $F$ records how dependent columns are formed,
- $P$ is a permutation matrix that restores the original column order.

#### Example 1

$
R = [I\quad  F] P= 
\begin{bmatrix}
1 & 0 & 3 & 5\\
0 & 1 & 4 & 6
\end{bmatrix}
$ The equation $Rx = 0$ gives $\begin{cases}x_1 + 3x_3 + 5x_4 = 0 \\x_2 + 4x_3 + 6x_4 = 0\end{cases}$

Here is a matrix $R$ of rank $r = 2$. It has $n = 4$ columns so we look for $n - r = 4 - 2 = 2$ independent solutions to $Rx = 0$. The nullspace $\mathcal{N}(R)$ will have dimension $2$.

Two **special solutions** are easy to find by setting the free variables $x_3, x_4$.

Case 1: $x_3 = 1,\; x_4 = 0$

- Equation 1 gives $x_1 = -3$
- Equation 2 gives $x_2 = -4$

$$
s_1 = (-3,\,-4,\,1,\,0)
$$

Case 2: $x_3 = 0,\; x_4 = 1$

- Equation 1 gives $x_1 = -5$
- Equation 2 gives $x_2 = -6$

$$
s_2 = (-5,\,-6,\,0,\,1)
$$

We can get the Nullspace $\mathcal{N}(R)$

The vectors
$$
s_1 = (-3,\,-4,\,1,\,0),
\quad
s_2 = (-5,\,-6,\,0,\,1)
$$
are in the **nullspace of $R$**:
$$
R s_1 = 0,
\qquad
R s_2 = 0.
$$

Any linear combination
$$
x = c_1 s_1 + c_2 s_2
$$
also satisfies $Rx = 0$.

Therefore, $s_1$ and $s_2$ form a **basis for the nullspace of $R$**.

**$R$ is an example of a matrix $A$ in “reduced row echelon form” (rref($A$)).** It has a more general case $R_0$, there can also be rows of zeros.

#### Example 2

$
R_0 =
\begin{bmatrix}
1 & 7 & 0 & 8\\
0 & 0 & 1 & 9\\
0 & 0 & 0 & 0
\end{bmatrix}
$ The equation $R_0 x = 0$ gives $\begin{cases}
x_1 + 7x_2 + 0x_3 + 8x_4 = 0 \\
x_3 + 9x_4 = 0 \\
0 = 0
\end{cases}
$

Now the identity matrix is inside columns 1 and 3. And row 3 is all zero. This still counts as a reduced row echelon form—elimination can’t make it simpler.

The special solutions still have 1 and 0 for the **free variables**—which are $x_2$ and $x_4$.

Case 1: $x_2 = 1,\; x_4 = 0$

- Equation 1 gives $x_1 = -7$
- Equation 2 gives $x_3 = 0$

$$
s_1 = (-7,\;1,\;0,\;0)
$$

Case 2: $x_2 = 0,\; x_4 = 1$

- Equation 1 gives $x_1 = -8$
- Equation 2 gives $x_3 = -9$

$$
s_2 = (-8,\;0,\;-9,\;1)
$$

The Nullspace of $R$:

The vectors
$
s_1 = (-7,\;1,\;0,\;0),
\quad
s_2 = (-8,\;0,\;-9,\;1)
$
form a **basis for the nullspace** of $R_0$.

If we want to keep $I$ in columns $1$ and $2$, we need a permutation $P$.


#### Take a summary:

1. **Simplest case**  

    $r,m,n = 2,2,4$

    $$
    R = [\, I \;\; F \,]P
    \quad \text{as in} \quad
    \begin{bmatrix}
    1 & 0 & 3 & 5\\
    0 & 1 & 4 & 6
    \end{bmatrix}
    $$

2. **General case**  
    
    $r,m,n = 2,3,4$

    $$
    R_0 =
    \begin{bmatrix}
    I & F\\
    0 & 0
    \end{bmatrix}
    P
    \quad \text{as in} \quad
    \begin{bmatrix}
    1 & 7 & 0 & 8\\
    0 & 0 & 1 & 9\\
    0 & 0 & 0 & 0
    \end{bmatrix}
    $$

#### Elimination from $A$ to rref($A$): Reduced Row Echelon Form

There are 3 different steps for elimination work:

1. Subtract a multiple of one row from another row (above or below !)
2. Multiply a row by any nonzero number
3. Exchange any rows.

##### Example 1:

Here is a 2 by 4 matrix $A$ that elimination reduces to our 2 by 4 example $R = [ I\quad F ]$:

$$A =
\begin{bmatrix}
1 & 2 & 11 & 17\\
3 & 7 & 37 & 57
\end{bmatrix}
\;\longrightarrow\;
\begin{bmatrix}
1 & 2 & 11 & 17\\
0 & 1 & 4 & 6
\end{bmatrix}
\;\longrightarrow\;
\begin{bmatrix}
1 & 0 & 3 & 5\\
0 & 1 & 4 & 6
\end{bmatrix}
= R$$

Elimination starts with column 1. It subtracts 3 times row 1 from row 2. That produces the zero in the middle matrix. Moving to column 2, we subtract 2 times the new row 2 from row 1. Upward elimination produces the second zero in $R$. Now $R$ starts with the $r$ by $r$ identity matrix $I$. The rank is $r = 2$ and elimination on this matrix $A$ is complete.

What did elimination actually do? **It inverted the leading 2 by 2 matrix $W=\begin{bmatrix}
1 & 2\\
3 & 7
\end{bmatrix}$**.

**Multiply $W^{-1}A = W^{-1}[\, W \;\; H \,]$ to produce $R = [\, I \;\; W^{-1}H \,] = [\, I \;\; F \,]$.**

The matrix $F$ is telling us how to combine the independent columns of $A$ to produce the dependent columns $H$.

**Dependent columns $H =\begin{bmatrix}
11 & 17\\
37 & 57
\end{bmatrix}=
W F=
\begin{bmatrix}
1 & 2\\
3 & 7
\end{bmatrix}
\begin{bmatrix}
3 & 5\\
4 & 6
\end{bmatrix}.
$**

**However you compute $R$ from $A$, you always reach the same $R$.** $R$ is completely determined by $A$ (even if there are different elimination steps that lead from $A$ to $R$).

$$\boxed{
\begin{aligned}
1.\;& \text{The first } r \text{ independent columns of } A \text{ locate the columns of } R \text{ containing } I.\\
   & \text{The permutation } P \text{ puts those columns into their correct places in } A.\\
2.\;& \text{The remaining columns } F \text{ in } R \text{ are determined by } H = WF:\\
   & (\text{Dependent columns of } A) = (\text{Independent columns of } A)\times F.\\
3.\;& \text{The last } m-r \text{ rows of } R_0 \text{ are rows of zeros.}
\end{aligned}
}$$


##### Example 2

Come back to the matrix $A$ that leads to our second reduced echelon form $R_0$. Both $A$ and $R_0$ are $3 \times 4$ matrices of rank $r = 2$. Watch each step of elimination:

$$
A =
\begin{bmatrix}
1 & 7 & 3 & 35\\
2 & 14 & 6 & 70\\
2 & 14 & 9 & 97
\end{bmatrix}
\;\longrightarrow\;
\begin{bmatrix}
1 & 7 & 3 & 35\\
0 & 0 & 0 & 0\\
0 & 0 & 3 & 27
\end{bmatrix}
\;\longrightarrow\;
\begin{bmatrix}
1 & 7 & 0 & 8\\
0 & 0 & 0 & 0\\
0 & 0 & 3 & 27
\end{bmatrix}
\;\longrightarrow\;
\begin{bmatrix}
1 & 7 & 0 & 8\\
0 & 0 & 1 & 9\\
0 & 0 & 0 & 0
\end{bmatrix}
= R_0
$$

- **$A \to B$:**  
  **(1)** Subtract multiples of row 1 from rows 2 and 3 to create zeros below the first pivot.

- **$B \to C$:**  
  **(1)** Subtract a multiple of row 3 from row 1 (upward elimination) to create a zero above the second pivot.

- **$C \to R_0$:**  
  **(2)** Multiply a row by a nonzero number to make the pivot equal to 1.  
  **(3)** Exchange rows to move the zero row to the bottom.


#### Elimination Column by Column: The Steps from $A$ to $R_0$

The algorithm performs elimination **one column at a time**, moving from **left to right**. After processing the first $k$ columns, that part of the matrix is already in **reduced row echelon form (rref)**, and the algorithm proceeds to column $k+1$.

$$\text{First } k+1 \text{ columns } \begin{bmatrix}
I_k & F_k \\
0   & 0
\end{bmatrix}
P_k$$

Each new column is split into:

$$\begin{bmatrix}
u \\
\ell
\end{bmatrix}$$

- an **upper part**, corresponding to rows already processed, and  
- a **lower part**, corresponding to rows not yet used.

The key question is whether the new column is **linearly dependent** on the first $k$ columns or **independent** of them:

- If the **lower part is all zeros**, the new column is dependent on the previous columns and becomes a **free (non-pivot) column**.
- If the **lower part contains a nonzero entry**, the new column is independent. A pivot is chosen from the lower part, row operations are used to eliminate the other entries in the column, and the column becomes a **pivot column**.

This process continues column by column until elimination is complete.

At the end, the algorithm identifies the **first $r$ independent columns of $A$**. These columns form the matrix $C$, and together with the row-reduced matrix $R$, give the factorization:

$$A=CR$$

- The matrix $R$ contains an $r × r$ identity matrix, revealing the rank and the independent columns of $A$.


##### Example 2: Row Operations from $A$ to $R_0$

Example 2 shows the **three allowed row operations** in elimination from $A$ to $R_0$:

1. **Subtract a multiple of one row from another row** (below or above)
2. **Divide a row by its first nonzero entry** (e.g. divide $[0\ 0\ 3\ 27]$ to reach pivot $=1$)
3. **Exchange rows** (to move all zero rows to the bottom of $R_0$)

A different series of steps could reach the same $R_0$, but the final result  
$$
R_0 = \operatorname{rref}(A)
$$
**cannot change**. All pieces of $R_0$ are completely determined by the original matrix $A$.

- $R_0$ has a **zero row** because $A$ has rank $r = 2$.
- The identity matrix $I$ appears in **columns 1 and 3** of $R_0$  
  because those are the **first independent columns** of $A$.
- The matrix $F$ appears in **columns 2 and 4** of $R$, combining  
  columns 1 and 3 of $A$ to produce the **dependent columns 2 and 4**.

$$
C F
=
\begin{bmatrix}
1 & 3 \\
2 & 6 \\
2 & 9
\end{bmatrix}
\begin{bmatrix}
7 & 8 \\
0 & 9
\end{bmatrix}
=
\begin{bmatrix}
7 & 35 \\
14 & 70 \\
14 & 97
\end{bmatrix}
$$

These are the **dependent columns 2 and 4 of $A$**.

We could never have seen in Chapter 1 that $(35, 70, 97)$ is a combination of columns 1 and 3 of $A$.

The matrix $R$ shows us the **nullspace of $A$**.

To solve:
$$
Ax = 0
$$
we only need to solve:
$$
Rx = 0
$$

This is easy because of the **identity matrix inside $R$**.


#### The Matrix Factorization $A = CR$ and the Nullspace

- There is a systematic way to find **independent columns**:

    1. Apply elimination to reduce $A$ to $R_0$.
    2. Then $I$ in $R_0$ locates the matrix $C$ of independent columns in $A$.
    3. Removing any zero rows from $R_0$ produces the row matrix $R$ in $A=CR$.

We find two special solutions $s_1$ and $s_2$ - one solution for every column of $F$ in $R$.

##### Example 2 (again)

$$

\begin{bmatrix}

\end{bmatrix}
\begin{bmatrix}

$$

Put **1** and **0** in positions **2** and **4**.

$$

\begin{bmatrix}

\end{bmatrix}
$$

Put **0** and **1** in positions **2** and **4**.

I think $s_1$ and $s_2$ are easiest to see using the matrices $-F$ and $I$ and $P^T$.

$$
\boxed{
\begin{aligned}
&\text{The two special solutions to } [\, I \;\; F \,]\,x = 0 
\text{ are the columns of }
\begin{bmatrix}

$$

Therefore, we can get a conclusion:

$$A = C \,[\, I \;\; F \,]\, P \;\text{multiplies}\;
X = P^{T}
\begin{bmatrix}
- F \\
I_{n-r}
\end{bmatrix}
\;\text{to give}\;
A X = 0.$$

Then the special solutions to $Ax = 0$ are the $n - r$ columns of $X$.

**Important.** Suppose $A$ has more columns than rows. With $n > m$ there is at least one free variable.  
The system $Ax = 0$ has at least one nonzero solution.

$$
\boxed{
\begin{aligned}
&\text{Suppose } Ax = 0 \text{ has more unknowns than equations } (n > m). \\
&\text{There must be at least } n - m \text{ free columns.} \\
&Ax = 0 \text{ must have nonzero solutions in } \mathcal{N}(A).
\end{aligned}
}
$$

The nullspace is a **subspace**. Its **dimension** is the number of free variables.  
This central idea—the **dimension of a subspace**—is explained in Section 3.5 of this chapter.

#### Block Elimination in Three Steps: Final Thoughts

If we work with blocks of the original $A$, then block elimination can be described in three quick steps. $A$ has rank $r$.

1. Step 1: Exchange columns of $A$ by $P_C$ and exchange rows of $A$ by $P_R$ to put the $r$ independent columns first and $r$ independent rows first in $P_R A P_C$.

    $P_R A P_C =
    \begin{bmatrix}
    W & H \\
    J & K
    \end{bmatrix},
    \qquad
    C =
    \begin{bmatrix}
    W \\
    J
    \end{bmatrix},
    \qquad
    B = [\, W \;\; H \,]
    \text{ have full rank } r.$
2. Step 2: Multiply the top $r$ rows by $W^{-1}$ to produce
    
    $
    W^{-1} B = [\, I \;\; W^{-1} H \,] = [\, I \;\; F \,].
    $
3. Step 3: Subtract $J[\, I \;\; W^{-1} H \,]$ from the lower $m-r$ rows $[\, J \;\; K \,]$ to produce $[\, 0 \;\; 0 \,]$

The result of Steps 1, 2, 3 is the reduced row echelon form $R_0$:

$$\boxed{
P_R A P_C
=
\begin{bmatrix}
W & H \\
J & K
\end{bmatrix}
\;\longrightarrow\;
\begin{bmatrix}
I & W^{-1} H \\
J & K
\end{bmatrix}
\;\longrightarrow\;
\begin{bmatrix}
I & W^{-1} H \\
0 & 0
\end{bmatrix}
= R_0
}
\tag{1}$$

There are two facts that need explanation. They lead to **Step 2** and **Step 3**:

1. The $r \times r$ matrix $W$ is **invertible**. 

    For the invertibility of $W$, we look back to the factorization $A = CR$. Focusing on the $r$ independent rows of $A$ that go into $B$, this is

    $
     B = W R.
    $

    Since $B$ and $R$ have rank $r$ and $W$ is $r \times r$, $W$ must have rank $r$ and therefore must be invertible.
2. The blocks satisfy $J W^{-1} H = K$.

    We know that the first $r$ rows $[\, I \;\; W^{-1}H \,]$ are linearly independent. Since $A$ has rank $r$, the lower rows $[\, J \;\; K \,]$ must be combinations of those upper rows. The combinations must be given by $J$ to get the first $r$ columns correct: $J I = J$. Then $J W^{-1} H$ must equal $K$ to make the last columns correct.

$$\boxed{\text{The conclusion is that }
P_R A P_C
=
\begin{bmatrix}
W \\
J
\end{bmatrix}
\, W^{-1} \,
[\, W \;\; H \,]
=
C \, W^{-1} \, B
}$$
